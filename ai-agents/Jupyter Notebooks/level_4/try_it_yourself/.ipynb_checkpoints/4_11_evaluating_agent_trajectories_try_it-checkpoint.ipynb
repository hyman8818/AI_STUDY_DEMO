````python
# Setup - run this cell first

import json
from dotenv import load_dotenv

from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import AIMessage, ToolMessage, HumanMessage

# Import agentevals components
from agentevals.trajectory.llm import (
    create_trajectory_llm_as_judge,
    TRAJECTORY_ACCURACY_PROMPT
)

# Load environment variables
load_dotenv()

print("Setup complete!")
````

## Book Lookup Tool and Agent

The book lookup tool and agent are provided for you. Your task is to evaluate the agent's trajectory, not to build the agent itself.

````python
# Mock book database
BOOK_DATABASE = {
    "1984": {
        "title": "1984",
        "author": "George Orwell",
        "year": 1949,
        "genre": "Dystopian Fiction",
        "summary": "A dystopian novel about totalitarian surveillance and control."
    },
    "dune": {
        "title": "Dune",
        "author": "Frank Herbert",
        "year": 1965,
        "genre": "Science Fiction",
        "summary": "An epic tale of politics, religion, and ecology on the desert planet Arrakis."
    },
    "pride and prejudice": {
        "title": "Pride and Prejudice",
        "author": "Jane Austen",
        "year": 1813,
        "genre": "Romance",
        "summary": "A witty exploration of love, class, and society in Regency-era England."
    }
}


@tool
def lookup_book(title: str) -> str:
    """Look up information about a book by its title.
    
    Args:
        title: The title of the book to look up
    
    Returns:
        A string with book information or an error message
    """
    title_key = title.lower().strip()
    
    if title_key in BOOK_DATABASE:
        book = BOOK_DATABASE[title_key]
        return (
            f"Title: {book['title']}\n"
            f"Author: {book['author']}\n"
            f"Year: {book['year']}\n"
            f"Genre: {book['genre']}\n"
            f"Summary: {book['summary']}"
        )
    
    return f"Book '{title}' not found in the database."


# Create the agent
agent_model = ChatOpenAI(model="gpt-4o-mini", temperature=0.1)
book_agent = create_agent(model=agent_model, tools=[lookup_book])

print("Book agent created!")
print(f"Available books: {', '.join(b['title'] for b in BOOK_DATABASE.values())}")
````

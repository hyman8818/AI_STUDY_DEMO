## Test Your Validation Model

Run these tests to verify your Pydantic model works correctly.

```python
# Test valid input
print("Testing valid inputs...")
try:
    query = MovieQuery(title="The Matrix")
    print(f"  Valid: '{query.title}'")
except Exception as e:
    print(f"  FAILED: {e}")

try:
    query = MovieQuery(title="  Inception  ")  # Should strip whitespace
    print(f"  Valid (stripped): '{query.title}'")
except Exception as e:
    print(f"  FAILED: {e}")

# Test invalid inputs
print("\nTesting invalid inputs (should fail)...")

try:
    query = MovieQuery(title="   ")  # Empty after strip
    print(f"  FAILED: Should have rejected whitespace-only title")
except ValueError as e:
    print(f"  Correctly rejected whitespace: {e}")

try:
    query = MovieQuery(title="Movie<script>alert('xss')</script>")  # Invalid chars
    print(f"  FAILED: Should have rejected special characters")
except ValueError as e:
    print(f"  Correctly rejected special chars: {e}")
```

## Create and Test the Agent

Create an agent with your movie tool and test it with natural language queries.

```python
# Create the agent
model = ChatOpenAI(model="gpt-4o-mini", temperature=0.1)
movie_agent = create_agent(model=model, tools=[get_movie_info])

print("Movie agent created successfully!")
```

```python
# Test the agent with a natural language query
query = "Tell me about the movie Inception. Who directed it?"
print(f"Query: {query}\n")

result = movie_agent.invoke({
    "messages": [{"role": "user", "content": query}]
})

print(f"Response: {result['messages'][-1].content}")
```

```python
# Test with another query
query = "What's The Godfather about and what year was it released?"
print(f"Query: {query}\n")

result = movie_agent.invoke({
    "messages": [{"role": "user", "content": query}]
})

print(f"Response: {result['messages'][-1].content}")
```

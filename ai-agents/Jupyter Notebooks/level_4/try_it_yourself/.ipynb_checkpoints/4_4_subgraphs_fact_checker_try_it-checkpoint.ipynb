# Subgraphs: Practice Exercise

Build a **Recipe Nutrition Analyzer** that uses nested subgraphs to extract ingredients from a recipe and analyze their nutritional properties.

**What you'll implement:**
- An ingredient extraction subgraph with two nodes
- A nutrition analysis subgraph that calls a nested allergen detection subgraph
- A main workflow that orchestrates the subgraphs

**Estimated time:** 15 minutes

## Setup

Run the setup cell to import dependencies and initialize the LLM.

````python
# Setup - run this cell first

from dotenv import load_dotenv
import os
from typing import TypedDict, List, Dict, Any
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage
from pydantic import BaseModel, Field

# Load environment variables
load_dotenv()
assert os.getenv("OPENAI_API_KEY"), "OPENAI_API_KEY not found"

# Initialize LLM
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# Pydantic models for structured output
class Ingredient(BaseModel):
    """A single ingredient from a recipe."""
    name: str = Field(description="Name of the ingredient")
    quantity: str = Field(description="Amount needed (e.g., '2 cups', '1 tbsp')")
    category: str = Field(description="Category: protein, vegetable, dairy, grain, spice, other")

class IngredientList(BaseModel):
    """List of ingredients extracted from a recipe."""
    ingredients: List[Ingredient] = Field(description="List of extracted ingredients")

class AllergenInfo(BaseModel):
    """Allergen information for an ingredient."""
    contains_gluten: bool = Field(description="Whether ingredient contains gluten")
    contains_dairy: bool = Field(description="Whether ingredient contains dairy")
    contains_nuts: bool = Field(description="Whether ingredient contains nuts")
    contains_soy: bool = Field(description="Whether ingredient contains soy")
    allergen_notes: str = Field(description="Additional allergen warnings")

class NutritionInfo(BaseModel):
    """Nutritional information for an ingredient."""
    ingredient: str = Field(description="Name of the ingredient")
    calories_per_serving: int = Field(description="Estimated calories")
    protein_grams: float = Field(description="Protein in grams")
    health_notes: str = Field(description="Brief health benefits or concerns")
    allergen_info: Dict[str, Any] = Field(description="Allergen details")

print("Setup complete!")
````

## Your Task

You will build a recipe analyzer with this architecture:

```
Main Workflow:
  Recipe Input -> Extract Ingredients -> Analyze Each Ingredient -> Generate Summary
                        |
                  Ingredient Extraction
                    Subgraph
                        |
                  parse -> categorize
                        
                                  Nutrition Analysis
                                    Subgraph (called per ingredient)
                                        |
                                  estimate_nutrition -> check_allergens
                                                              |
                                                        Allergen Detection
                                                          Subgraph
```

**Input:** A recipe with ingredients list

**Output:** A nutrition summary with allergen warnings for each ingredient

## Step 1: Define State Schemas

Define the state schemas for each level of the workflow. Each subgraph has its own isolated state.

````python
# State schema for the main workflow
class MainRecipeState(TypedDict):
    """State for the main recipe analysis workflow."""
    recipe_text: str
    ingredients: List[Dict[str, Any]]
    nutrition_results: List[Dict[str, Any]]
    final_summary: str

# State schema for ingredient extraction subgraph
class IngredientExtractionState(TypedDict):
    """State for ingredient extraction subgraph."""
    recipe_text: str
    parsed_ingredients: str
    categorized_ingredients: List[Dict[str, Any]]

# State schema for nutrition analysis subgraph
class NutritionAnalysisState(TypedDict):
    """State for nutrition analysis subgraph."""
    ingredient_name: str
    ingredient_quantity: str
    nutrition_estimate: Dict[str, Any]
    allergen_assessment: Dict[str, Any]

# State schema for allergen detection subgraph
class AllergenDetectionState(TypedDict):
    """State for allergen detection subgraph."""
    ingredient_name: str
    allergen_result: Dict[str, Any]

print("State schemas defined!")
````

## Step 3: Build the Nutrition Analysis Subgraph

This subgraph estimates nutrition and calls the allergen subgraph. It will be invoked once per ingredient.

````python
def estimate_nutrition_node(state: NutritionAnalysisState) -> NutritionAnalysisState:
    """
    Estimates nutritional information for an ingredient.
    
    Args:
        state: Contains ingredient_name and ingredient_quantity
    
    Returns:
        Updated state with nutrition_estimate containing:
        - calories_per_serving (int)
        - protein_grams (float)
        - health_notes (str)
    """
    ingredient = state["ingredient_name"]
    quantity = state["ingredient_quantity"]
    
    print(f"  Estimating nutrition for: {quantity} {ingredient}")
    
    system_prompt = """You are a nutritionist. Estimate the nutritional content for the given ingredient and quantity.
    Provide realistic estimates for calories and protein, plus brief health notes."""
    
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": f"Ingredient: {quantity} {ingredient}"}
    ]
    
    # Define a simple model for nutrition estimate
    class NutritionEstimate(BaseModel):
        calories_per_serving: int
        protein_grams: float
        health_notes: str
    
    estimator = llm.with_structured_output(NutritionEstimate)
    result = estimator.invoke(messages)
    
    return {**state, "nutrition_estimate": result.model_dump()}


def check_allergens_node(state: NutritionAnalysisState) -> NutritionAnalysisState:
    """
    Checks for allergens by invoking the allergen detection subgraph.
    
    Args:
        state: Contains ingredient_name to check
    
    Returns:
        Updated state with allergen_assessment populated from subgraph result
    """
    # TODO: Implement allergen checking by invoking the allergen_subgraph
    # 1. Create subgraph_input dict with ingredient_name and empty allergen_result
    # 2. Invoke allergen_subgraph with the input
    # 3. Extract allergen_result from subgraph output
    # 4. Return state with allergen_assessment set to the result
    pass


# Build the nutrition analysis subgraph
nutrition_builder = StateGraph(NutritionAnalysisState)
nutrition_builder.add_node("estimate_nutrition", estimate_nutrition_node)
nutrition_builder.add_node("check_allergens", check_allergens_node)
nutrition_builder.add_edge(START, "estimate_nutrition")
nutrition_builder.add_edge("estimate_nutrition", "check_allergens")
nutrition_builder.add_edge("check_allergens", END)
nutrition_subgraph = nutrition_builder.compile()

print("Nutrition Analysis Subgraph built!")
````

## Step 4: Build the Ingredient Extraction Subgraph

This subgraph parses the recipe and extracts/categorizes ingredients.

````python
def parse_ingredients_node(state: IngredientExtractionState) -> IngredientExtractionState:
    """
    Parses the recipe text and extracts ingredients using structured output.
    
    Args:
        state: Contains recipe_text to parse
    
    Returns:
        Updated state with parsed_ingredients as a string representation
    """
    recipe = state["recipe_text"]
    
    print("Parsing recipe for ingredients...")
    
    system_prompt = """You are a culinary expert. Extract all ingredients from the recipe.
    For each ingredient, identify:
    - The name of the ingredient
    - The quantity needed
    - The category (protein, vegetable, dairy, grain, spice, other)"""
    
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": f"Recipe:\n{recipe}"}
    ]
    
    extractor = llm.with_structured_output(IngredientList)
    result = extractor.invoke(messages)
    
    print(f"Found {len(result.ingredients)} ingredients")
    
    return {**state, "parsed_ingredients": str(result.model_dump())}


def categorize_ingredients_node(state: IngredientExtractionState) -> IngredientExtractionState:
    """
    Converts parsed ingredients string to a list of dicts and sorts by category.
    
    Args:
        state: Contains parsed_ingredients as string
    
    Returns:
        Updated state with categorized_ingredients as list of dicts
    """
    parsed = eval(state["parsed_ingredients"])
    ingredients_list = parsed["ingredients"]
    
    # Sort by category for organized output
    sorted_ingredients = sorted(ingredients_list, key=lambda x: x["category"])
    
    print("Ingredients categorized:")
    for ing in sorted_ingredients:
        print(f"  [{ing['category']}] {ing['quantity']} {ing['name']}")
    
    return {**state, "categorized_ingredients": sorted_ingredients}


# Build the ingredient extraction subgraph
extraction_builder = StateGraph(IngredientExtractionState)
extraction_builder.add_node("parse_ingredients", parse_ingredients_node)
extraction_builder.add_node("categorize_ingredients", categorize_ingredients_node)
extraction_builder.add_edge(START, "parse_ingredients")
extraction_builder.add_edge("parse_ingredients", "categorize_ingredients")
extraction_builder.add_edge("categorize_ingredients", END)
extraction_subgraph = extraction_builder.compile()

print("\nIngredient Extraction Subgraph built!")
````

## Step 5: Build the Main Workflow

The main workflow orchestrates everything: extract ingredients, analyze each one, generate summary.

````python
def extract_ingredients_main(state: MainRecipeState) -> MainRecipeState:
    """
    Extracts ingredients by invoking the extraction subgraph.
    
    Args:
        state: Contains recipe_text
    
    Returns:
        Updated state with ingredients list populated
    """
    # TODO: Implement ingredient extraction by invoking extraction_subgraph
    # 1. Create subgraph_input with recipe_text, empty parsed_ingredients, empty categorized_ingredients
    # 2. Invoke extraction_subgraph
    # 3. Extract categorized_ingredients from output
    # 4. Return state with ingredients set to the result
    pass


def analyze_nutrition_main(state: MainRecipeState) -> MainRecipeState:
    """
    Analyzes nutrition for each ingredient by invoking nutrition_subgraph multiple times.
    
    Args:
        state: Contains ingredients list
    
    Returns:
        Updated state with nutrition_results for all ingredients
    """
    ingredients = state["ingredients"]
    
    print("\n" + "="*60)
    print("ANALYZING NUTRITION FOR EACH INGREDIENT")
    print("="*60)
    
    nutrition_results = []
    
    # Limit to first 3 ingredients to keep exercise manageable
    for i, ingredient in enumerate(ingredients[:3], 1):
        print(f"\n--- Ingredient {i}/3 ---")
        
        # Transform to subgraph state
        subgraph_input = {
            "ingredient_name": ingredient["name"],
            "ingredient_quantity": ingredient["quantity"],
            "nutrition_estimate": {},
            "allergen_assessment": {}
        }
        
        # Invoke nutrition subgraph
        subgraph_output = nutrition_subgraph.invoke(subgraph_input)
        
        # Collect results
        nutrition_results.append({
            "ingredient": ingredient["name"],
            "quantity": ingredient["quantity"],
            "nutrition": subgraph_output["nutrition_estimate"],
            "allergens": subgraph_output["allergen_assessment"]
        })
    
    return {**state, "nutrition_results": nutrition_results}


def generate_summary_main(state: MainRecipeState) -> MainRecipeState:
    """
    Generates a final nutrition and allergen summary.
    
    Args:
        state: Contains nutrition_results
    
    Returns:
        Updated state with final_summary
    """
    import json
    
    results = state["nutrition_results"]
    
    print("\n" + "="*60)
    print("GENERATING SUMMARY")
    print("="*60)
    
    system_prompt = """You are a nutritionist creating a summary report.
    Summarize the nutrition information and highlight any allergen warnings.
    Be concise but thorough."""
    
    messages = [
        SystemMessage(content=system_prompt),
        HumanMessage(content=f"Nutrition data:\n{json.dumps(results, indent=2)}")
    ]
    
    response = llm.invoke(messages)
    
    return {**state, "final_summary": response.content}


# Build the main workflow
main_builder = StateGraph(MainRecipeState)
main_builder.add_node("extract_ingredients", extract_ingredients_main)
main_builder.add_node("analyze_nutrition", analyze_nutrition_main)
main_builder.add_node("generate_summary", generate_summary_main)
main_builder.add_edge(START, "extract_ingredients")
main_builder.add_edge("extract_ingredients", "analyze_nutrition")
main_builder.add_edge("analyze_nutrition", "generate_summary")
main_builder.add_edge("generate_summary", END)

recipe_analyzer = main_builder.compile()

print("Main Workflow built!")
print("\nArchitecture:")
print("  extract_ingredients -> analyze_nutrition -> generate_summary")
print("\nSubgraph relationships:")
print("  - extract_ingredients invokes: extraction_subgraph")
print("  - analyze_nutrition invokes: nutrition_subgraph (multiple times)")
print("  - nutrition_subgraph invokes: allergen_subgraph")
````

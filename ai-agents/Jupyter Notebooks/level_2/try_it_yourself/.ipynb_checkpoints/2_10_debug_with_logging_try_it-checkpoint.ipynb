## Agent and Tools (Provided)

The Agent class and tools are provided for you. Review them to understand what data is available for logging.

```python
class Agent:
    """Agent class that handles OpenAI API interactions."""
    
    def __init__(self, system=""):
        self.system = system
        self.messages = []
        if self.system:
            self.messages.append({"role": "system", "content": system})
    
    def __call__(self, message):
        self.messages.append({"role": "user", "content": message})
        result = self.execute()
        self.messages.append({"role": "assistant", "content": result})
        return result
    
    def execute(self):
        completion = client.chat.completions.create(
            model="gpt-4o",
            temperature=0,
            messages=self.messages
        )
        return completion.choices[0].message.content


# Library system tools
def get_book_info(title):
    """Get information about a book."""
    books = {
        "1984": {"author": "George Orwell", "year": 1949, "genre": "Dystopian"},
        "dune": {"author": "Frank Herbert", "year": 1965, "genre": "Science Fiction"},
        "pride and prejudice": {"author": "Jane Austen", "year": 1813, "genre": "Romance"},
    }
    return books.get(title.lower(), "Book not found")


def check_availability(title):
    """Check if a book is available in the library."""
    availability = {
        "1984": True,
        "dune": False,
        "pride and prejudice": True,
    }
    available = availability.get(title.lower())
    if available is None:
        return "Book not in catalog"
    return "Available" if available else "Checked out"


# System prompt for the agent
system_prompt = """
You run in a loop of Thought, Action, PAUSE, Observation.
At the end of the loop you output an Answer.
Use Thought to describe your thoughts about the question.
Use Action to run one of the actions available to you - then return PAUSE.
Observation will be the result of running those actions.

Your available actions are:

get_book_info:
e.g. get_book_info: 1984
Returns information about a book (author, year, genre)

check_availability:
e.g. check_availability: 1984
Returns whether the book is available or checked out

Example session:

Question: Who wrote Dune and is it available?
Thought: I need to get the book info for Dune first
Action: get_book_info: Dune
PAUSE

Observation: {'author': 'Frank Herbert', 'year': 1965, 'genre': 'Science Fiction'}

Thought: Now I need to check if Dune is available
Action: check_availability: Dune
PAUSE

Observation: Checked out

Answer: Dune was written by Frank Herbert and is currently checked out.
""".strip()

# Action parsing regex and known actions
action_re = re.compile(r'^Action: (\w+): (.*)$', re.MULTILINE)
known_actions = {
    "get_book_info": get_book_info,
    "check_availability": check_availability
}

print("Agent and tools ready.")
```

## Part 2: Add Logging to the Query Function

Complete the query function by adding logging calls at the marked locations.

```python
def query(question: str, max_turns: int = 5) -> None:
    """Run agent loop with logging at key decision points.
    
    Should log three types of events:
    1. 'query_started' at the beginning with {"question": ..., "max_turns": ...}
    2. 'tool_executed' after each tool runs with {"action": ..., "input": ..., "result": ...}
    3. 'query_completed' at the end with {"total_turns": ...}
    
    Args:
        question: User's question to the agent
        max_turns: Maximum number of agent turns allowed
    
    Returns:
        None (prints output and logs events)
    """
    # TODO 1: Log query_started event with question and max_turns
    
    agent = Agent(system_prompt)
    next_prompt = question
    i = 0
    
    while i < max_turns:
        i += 1
        result = agent(next_prompt)
        print(f"\n=== Turn {i}/{max_turns} ===")
        print(result)
        
        actions = [
            action_re.match(a)
            for a in result.split('\n')
            if action_re.match(a)
        ]
        
        if actions:
            action, action_input = actions[0].groups()
            
            if action not in known_actions:
                raise Exception(f"Unknown action: {action}: {action_input}")
            
            print(f"\n-- Running {action} with input: {action_input}")
            observation = known_actions[action](action_input)
            print(f"Observation: {observation}")
            
            # TODO 2: Log tool_executed event with action, input, and result
            
            next_prompt = f"Observation: {observation}"
        else:
            # TODO 3: Log query_completed event with total_turns
            
            return


print("Query function defined.")
```

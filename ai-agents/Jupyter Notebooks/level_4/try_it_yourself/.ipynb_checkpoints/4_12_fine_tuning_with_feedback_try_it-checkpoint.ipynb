## Setup

Run this cell to import all required libraries and configure the environment.

```python
# Setup - run this cell first

import os
from datetime import datetime
from typing import Dict, Any

from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool

# Load environment variables
load_dotenv()

if not os.getenv("OPENAI_API_KEY"):
    raise ValueError("OPENAI_API_KEY not found in environment variables")

print("Setup complete!")
```

## Context

You are improving a recipe assistant that helps users find and understand recipes. The current agent has a basic system prompt and users are providing feedback about its responses. Your task is to implement the meta-prompting functions that will analyze this feedback and generate improved instructions.

**Scenario:** A user asked the recipe assistant for a pasta recipe. The agent gave a correct but unhelpful response. The user rated it 2/5 and left specific feedback about what was missing.

**Your task:** Implement two functions:
1. `build_optimization_prompt` - Creates the meta-prompt that instructs an LLM to analyze the feedback and write a better system prompt
2. `create_improved_instructions` - Uses the meta-prompt with an LLM to generate the improved system prompt

**Key insight:** The meta-prompt acts as instructions for a "prompt engineer LLM" - it tells the LLM what information to analyze and how to write better instructions.

## Provided Components

The recipe tool, agent creation, and feedback data are provided for you.

```python
# Recipe tool - provided
@tool
def get_recipe(dish_name: str) -> str:
    """Get a recipe for a specific dish.
    
    Args:
        dish_name: The name of the dish to get a recipe for
        
    Returns:
        A string containing the recipe details
    """
    recipes = {
        "spaghetti carbonara": "Ingredients: pasta, eggs, pecorino, guanciale, pepper. Cook pasta, fry guanciale, mix eggs with cheese, combine all off heat.",
        "chicken stir fry": "Ingredients: chicken breast, vegetables, soy sauce, garlic, ginger. Slice chicken, stir fry with vegetables, add sauce.",
        "caesar salad": "Ingredients: romaine lettuce, croutons, parmesan, caesar dressing. Chop lettuce, add toppings, drizzle dressing.",
        "tomato soup": "Ingredients: tomatoes, onion, garlic, cream, basil. Saute onion and garlic, add tomatoes, blend, add cream.",
        "pancakes": "Ingredients: flour, eggs, milk, butter, sugar. Mix batter, cook on griddle until golden."
    }
    
    normalized_dish = dish_name.lower().strip()
    
    if normalized_dish in recipes:
        return recipes[normalized_dish]
    else:
        return f"Recipe not found for {dish_name}. Try: spaghetti carbonara, chicken stir fry, caesar salad, tomato soup, or pancakes."


def create_recipe_agent(system_prompt: str):
    """Create a recipe agent with the given system prompt."""
    model = ChatOpenAI(model="gpt-4o", temperature=0.1)
    agent = create_agent(model=model, tools=[get_recipe])
    return agent, system_prompt


def query_agent(agent, system_prompt: str, question: str) -> str:
    """Ask the agent a question and return the response."""
    result = agent.invoke({
        "messages": [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": question}
        ]
    })
    return result['messages'][-1].content


print("Recipe agent components ready!")
```

```python
# Current system prompt and feedback data - provided
BASELINE_PROMPT = "You are a recipe assistant."

# Simulated interaction and feedback
user_feedback = {
    "query": "How do I make spaghetti carbonara?",
    "response": "Here's the recipe for spaghetti carbonara: Ingredients: pasta, eggs, pecorino, guanciale, pepper. Cook pasta, fry guanciale, mix eggs with cheese, combine all off heat.",
    "rating": 2,
    "comments": (
        "The response just lists ingredients and steps without any detail. "
        "I need specific quantities and measurements. "
        "Also, I'd like cooking tips to avoid common mistakes like scrambling the eggs. "
        "The response feels robotic - a friendlier tone would be nice."
    ),
    "timestamp": datetime.now().isoformat()
}

print("Baseline prompt:")
print(f"  '{BASELINE_PROMPT}'")
print(f"\nUser feedback (rating {user_feedback['rating']}/5):")
print(f"  {user_feedback['comments']}")
```

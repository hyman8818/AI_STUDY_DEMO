## Setup

Run this cell to load dependencies and create the stock tools you'll use.

```python
# Setup - run this cell first

from dotenv import load_dotenv
load_dotenv()

from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool


@tool
def get_stock_price(symbol: str) -> str:
    """Get the current stock price for a given ticker symbol.
    
    Args:
        symbol: The stock ticker symbol (e.g., AAPL, GOOGL)
    
    Returns:
        A string with the current price, or a message if not available
    """
    # Mock implementation with limited data
    stock_prices = {
        "aapl": "$178.50",
        "googl": "$141.25",
        "msft": "$378.90",
        "amzn": "$178.15",
    }
    
    symbol_key = symbol.lower().strip()
    
    if symbol_key in stock_prices:
        return f"Current price of {symbol.upper()}: {stock_prices[symbol_key]}"
    else:
        return f"Price data not available for {symbol.upper()}"


@tool
def get_company_info(symbol: str) -> str:
    """Get basic company information for a given ticker symbol.
    
    Args:
        symbol: The stock ticker symbol (e.g., AAPL, GOOGL)
    
    Returns:
        A string with company information, or a message if not available
    """
    # Mock implementation with limited data
    company_info = {
        "aapl": "Apple Inc. - Technology sector, Market Cap: $2.8T",
        "googl": "Alphabet Inc. - Technology sector, Market Cap: $1.7T",
        "msft": "Microsoft Corp. - Technology sector, Market Cap: $2.9T",
        "amzn": "Amazon.com Inc. - Consumer Cyclical sector, Market Cap: $1.8T",
    }
    
    symbol_key = symbol.lower().strip()
    
    if symbol_key in company_info:
        return f"{symbol.upper()}: {company_info[symbol_key]}"
    else:
        return f"Company information not available for {symbol.upper()}"


stock_tools = [get_stock_price, get_company_info]


# Configure the language model
model = ChatOpenAI(
    model="gpt-4o",
    temperature=0.1
)


def query_agent(agent, question: str):
    """Query the agent and display tool calls and response.
    
    Args:
        agent: The LangChain agent to query
        question: The question to ask
    """
    print(f"\n{'='*60}")
    print(f"Question: {question}")
    print(f"{'='*60}\n")
    
    for chunk in agent.stream(
        {"messages": [{"role": "user", "content": question}]},
        stream_mode="updates"
    ):
        if "model" in chunk:
            messages = chunk["model"].get("messages", [])
            for msg in messages:
                if hasattr(msg, 'tool_calls') and msg.tool_calls:
                    for tool_call in msg.tool_calls:
                        print(f"Tool Call: {tool_call['name']}")
                        print(f"   Input: {tool_call['args']}")
                elif hasattr(msg, 'content') and msg.content:
                    print(f"\nAgent Response:\n{msg.content}")
        
        if "tools" in chunk:
            messages = chunk["tools"].get("messages", [])
            for msg in messages:
                if hasattr(msg, 'content'):
                    print(f"   Result: {msg.content}\n")
    
    print(f"\n{'='*60}\n")


print("Setup complete!")
print(f"Available tools: {[t.name for t in stock_tools]}")
```

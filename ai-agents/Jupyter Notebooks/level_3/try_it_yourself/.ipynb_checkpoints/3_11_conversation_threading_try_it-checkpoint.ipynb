## Run Your Implementation

Test your implementation by having separate conversations about two different projects.

```python
# Project 1: Website Redesign
website_config = {"configurable": {"thread_id": "project_website_redesign"}}

print("=" * 60)
print("PROJECT: Website Redesign")
print("=" * 60)

print("\nUser: We're using React and TypeScript for the website redesign.")
result = app.invoke(
    {"messages": [HumanMessage(content="We're using React and TypeScript for the website redesign.")]},
    config=website_config
)
print(f"Assistant: {result['messages'][-1].content}")

print("\nUser: The deadline is March 15th.")
result = app.invoke(
    {"messages": [HumanMessage(content="The deadline is March 15th.")]},
    config=website_config
)
print(f"Assistant: {result['messages'][-1].content}")
```

```python
# Project 2: Mobile App
mobile_config = {"configurable": {"thread_id": "project_mobile_app"}}

print("=" * 60)
print("PROJECT: Mobile App")
print("=" * 60)

print("\nUser: The mobile app is being built with Flutter and Dart.")
result = app.invoke(
    {"messages": [HumanMessage(content="The mobile app is being built with Flutter and Dart.")]},
    config=mobile_config
)
print(f"Assistant: {result['messages'][-1].content}")

print("\nUser: We need to launch by Q2.")
result = app.invoke(
    {"messages": [HumanMessage(content="We need to launch by Q2.")]},
    config=mobile_config
)
print(f"Assistant: {result['messages'][-1].content}")
```

```python
# Verify thread isolation - ask about tech stack in each project
print("=" * 60)
print("VERIFYING THREAD ISOLATION")
print("=" * 60)

print("\n--- Website Redesign Thread ---")
print("User: What tech stack are we using for this project?")
result = app.invoke(
    {"messages": [HumanMessage(content="What tech stack are we using for this project?")]},
    config=website_config
)
print(f"Assistant: {result['messages'][-1].content}")

print("\n--- Mobile App Thread ---")
print("User: What tech stack are we using for this project?")
result = app.invoke(
    {"messages": [HumanMessage(content="What tech stack are we using for this project?")]},
    config=mobile_config
)
print(f"Assistant: {result['messages'][-1].content}")

print("\n" + "=" * 60)
print("Success! Each project thread remembers its own context.")
print("=" * 60)
```
